```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


df = pd.read_csv('/content/parkinsons.data')
print("Dataset loaded successfully!")

print("--- FIRST 5 ROWS ---")
display(df.head())

print("--- LAST 5 ROWS ---")
display(df.tail())

print(f"\nDataset Shape (Rows, Columns): {df.shape}\n")

print("--- DATASET INFO ---")
df.info()

print("\n--- SUMMARY STATISTICS ---")
display(df.describe())

print("--- MISSING VALUES PER COLUMN ---")
print(df.isnull().sum())

print(f"\nTotal Duplicate Rows: {df.duplicated().sum()}")
# questions 1-4
print("--- (1) RECORD COUNTS BY STATUS ---")
status_counts = df['status'].value_counts()
print(status_counts)

# (2) Mean, median, min, max using NumPy arrays & Pandas
selected_cols = ['MDVP:Fo(Hz)', 'MDVP:Jitter(%)', 'MDVP:Shimmer']
print("\n--- (2) SUMMARY STATS FOR SELECTED MEASUREMENTS (PANDAS) ---")
display(df[selected_cols].agg(['mean', 'median', 'min', 'max']))

# Explicit NumPy array calculation (to fulfill the NumPy array requirement)
fo_array = df['MDVP:Fo(Hz)'].to_numpy()
print(f"NumPy Check for MDVP:Fo(Hz) -> Mean: {np.mean(fo_array):.2f}, "
      f"Median: {np.median(fo_array):.2f}, Min: {np.min(fo_array):.2f}, Max: {np.max(fo_array):.2f}")

# (3) Compare average of these measurements by status
print("\n--- (3) AVERAGE MEASUREMENTS BY STATUS ---")
display(df.groupby('status')[selected_cols].mean())

# (4) Top 10 highest values for one selected measurement
print("\n--- (4) TOP 10 HIGHEST MDVP:Fo(Hz) VALUES ---")
top_10_fo = df[['name', 'status', 'MDVP:Fo(Hz)']].sort_values(by='MDVP:Fo(Hz)', ascending=False).head(10)
display(top_10_fo)

# Pandas filtering example (required by prompt)
high_fo_records = df[df['MDVP:Fo(Hz)'] > 200]
print(f"\nFilter Check: Number of records with MDVP:Fo(Hz) > 200 Hz: {len(high_fo_records)}")

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Chart 1: Status-Count Bar Chart
status_counts.plot(kind='bar', ax=axes[0], color=['steelblue', 'coral'], edgecolor='black')
axes[0].set_title('Record Count by Status')
axes[0].set_xlabel('Status')
axes[0].set_ylabel('Number of Records')
axes[0].tick_params(axis='x', rotation=0)

# Chart 2: Histogram of MDVP:Fo(Hz)
axes[1].hist(df['MDVP:Fo(Hz)'], bins=15, color='mediumseagreen', edgecolor='black')
axes[1].set_title('Distribution of MDVP:Fo(Hz)')
axes[1].set_xlabel('MDVP:Fo(Hz)')
axes[1].set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```
